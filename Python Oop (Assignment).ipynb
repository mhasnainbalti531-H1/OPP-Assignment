### **Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self,amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print(f"Deposit Successfully your New Balance is {self.__balance}")
        else:
            print("Invalid Amount")
    def withdraw(self,amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance = self.__balance - amount
            print(f"Withdraw Successfully Your New Balance is {self.__balance}")
        else:
            print("insufficient Balance")
```

```python
b1=BankAccount(6790)
```

```python
b1.deposit(5000)
```

```python
b1.withdraw(1000) 
```

```python
b1.withdraw(10000)
```

```python
 b1.withdraw(600000)
```

### **Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
class Animal:
    def __init__(self,name):
        self.name = name
    def eat(self):
            return f"{self.name} is eating food"
class Mammal(Animal):
    def __init__(self,name,cast):
        super().__init__(name)
        self.cast = cast
    def warm_blooded(self):
        return f"{self.cast} is warm blooded"
class Dog(Mammal):
    def __init__(self,name,cast,dog_name):
        self.dog_name = dog_name
        super().__init__(name,cast)
    def bark(self):
        return f"{self.dog_name} is barking"
```

```python
d1=Dog("mama","camel","german")
```

```python
d1.bark()
```

```python
d1.warm_blooded()
```

```python
d1.eat()
```

### **Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
class Father:
    def __init__(self,father_skill):
        self.father_skill = father_skill
    def show_fatherskills(self):
        return self.father_skill
class Mother:
    def __init__(self,mother_skill):
        self.mother_skill = mother_skill
    def show_motherskills(self):
        return self.mother_skill
class Child(Father,Mother):
    def __init__(self,father_skill,mother_skill,child_skills):
        Father.__init__(self,father_skill)
        Mother.__init__(self,mother_skill)
        self.child_skills = child_skills
    def show_childskills(self):
        return self.child_skills
```

```python
c1=Child("finance","cooking","trading")
```

```python
c1.show_childskills()
```

```python
c1.show_fatherskills()
```

```python
c1.show_motherskills()
```

### **Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def show_employeeinfo(self):
        print(f"""
Name   : {self.name}
Salary : {self.salary}""")
class Manager(Employee):
    def __init__(self,name,salary,department):
        self.department = department
        super().__init__(name,salary)  # super() parent class ka costructor/init ko call ke liye use kar tte hai
    def show_managerinfo(self):
        print(f"""
Name        :  {self.name}  
Salary      :  {self.salary}
Department  :  {self.department}""")
```

```python
m1=Manager("Hasnain",8000,"ai")
```

```python
m1.show_managerinfo()
```

```python
m1.show_employeeinfo()
```

### **Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
#jab aik parent class sa 2 ya 2 sa zyada child classes inhrit karte hai us se "Hierarchical inheritance bolte hai"
class Shape:
    def __init__(self,color):
        self.color = color
    def show_shapecolor(self):
        return f"The Shape color is {self.color}"
class Circle(Shape):
    def __init__(self,color,radius):
        self.radius = radius
        super().__init__(color)
    def show_circledetails(self):
        print(f"""
Circle Color   :  {self.color}
Circle Radius  :  {self.radius}""")
class Square(Shape):
    def __init__(self,color,sides):
        self.sides = sides
        super().__init__(color)
    def show_squaredetails(self):
        print(f"""
Square Colour :  {self.color}
Square Sides  :  {self.sides}""")
```

```python
s1=Square("red",6)
s2=Square("black",89)
```

```python
s1.show_squaredetails()
```

```python
s1.show_shapecolor()
```

```python
s2.show_squaredetails()
```

```python
s2.show_shapecolor()
```

### **Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

### **Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Book:
    def __init__(self,pages):
        self.pages = pages
    def __add__(self,others):
        return self.pages + others.pages
```

```python
book1=Book(500)
book2=Book(1000)
print(book1+book2)
```

### **Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC , abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class CreditCardPayment(Payment):
    def pay(self,amount):
        print(f"Paid RS: {amount} using CreditCard")
class UPIPayment(Payment):
    def pay(self,amount):
        print(f"Paid RS: {amount} using UPI Payment")
```

```python
p1=CreditCardPayment()
p2=UPIPayment()
```

```python
p1.pay(100)
```

```python
p2.pay(700)
```
